```python
import pandas as pd
import numpy as np
import os
import glob

# Project folder
base_path = "/content/drive/MyDrive/RareTumor-Omics/Raw_Datasets"

# Find dataset files
mutation_file = glob.glob(base_path + "/Genomics/*.cbt")[0]

methylation_file = glob.glob(base_path + "/Epigenomics/*.cct.gz")[0]

rnaseq_file = glob.glob(base_path + "/Transcriptomics/*.cct.gz")[0]

clinical_file = glob.glob(base_path + "/Clinical/*.tsi")[0]

# Load datasets
genomics = pd.read_csv(
    mutation_file, sep="\t", index_col=0
)

methylation = pd.read_csv(
    methylation_file, sep="\t", index_col=0,
    compression="gzip", na_values=["NA"]
)

transcriptomics = pd.read_csv(
    rnaseq_file, sep="\t", index_col=0,
    compression="gzip", na_values=["NA"]
)

clinical = pd.read_csv(
    clinical_file, sep="\t", index_col=0,
    na_values=["NA"]
)

print("All four datasets loaded successfully!")
```

```python
print("GENOMICS:", genomics.shape)

print("METHYLATION:", methylation.shape)

print("TRANSCRIPTOMICS:", transcriptomics.shape)

print("CLINICAL:", clinical.shape)
```

```python
# Transpose all four datasets

genomics_T = genomics.T
methylation_T = methylation.T
transcriptomics_T = transcriptomics.T
clinical_T = clinical.T

print("Datasets transposed successfully!")

print("Genomics:", genomics_T.shape)
print("Methylation:", methylation_T.shape)
print("Transcriptomics:", transcriptomics_T.shape)
print("Clinical:", clinical_T.shape)
```

```python
# Check the number of unique patient samples in each dataset

print("Genomics samples:", len(genomics_T.index.unique()))

print("Methylation samples:", len(methylation_T.index.unique()))

print("Transcriptomics samples:", len(transcriptomics_T.index.unique()))

print("Clinical samples:", len(clinical_T.index.unique()))


# Find samples common to all four datasets

common_samples = (
    set(genomics_T.index)
    & set(methylation_T.index)
    & set(transcriptomics_T.index)
    & set(clinical_T.index)
)

print("\nCommon samples across all four datasets:", len(common_samples))
```

```python
# Arrange all datasets using the same patient sample order

common_samples = sorted(
    set(genomics_T.index)
    & set(methylation_T.index)
    & set(transcriptomics_T.index)
    & set(clinical_T.index)
)

# Reorder all datasets

genomics_T = genomics_T.loc[common_samples]
methylation_T = methylation_T.loc[common_samples]
transcriptomics_T = transcriptomics_T.loc[common_samples]
clinical_T = clinical_T.loc[common_samples]

print("All datasets aligned successfully!")

# Verify sample order

print("Genomics and Methylation:",
      (genomics_T.index == methylation_T.index).all())

print("Genomics and Transcriptomics:",
      (genomics_T.index == transcriptomics_T.index).all())

print("Genomics and Clinical:",
      (genomics_T.index == clinical_T.index).all())
```

```python
# Count missing values in each dataset

datasets = {
    "Genomics": genomics_T,
    "Methylation": methylation_T,
    "Transcriptomics": transcriptomics_T,
    "Clinical": clinical_T
}

for name, df in datasets.items():

    total_missing = df.isnull().sum().sum()
    total_values = df.size
    missing_percentage = (total_missing / total_values) * 100

    print("\n", name)
    print("Total values:", total_values)
    print("Missing values:", total_missing)
    print("Missing percentage:", round(missing_percentage, 2), "%")
```

```python
# Identify features containing missing values

print("METHYLATION FEATURES WITH MISSING VALUES")

methylation_missing = methylation_T.isnull().sum()
print(methylation_missing[methylation_missing > 0])


print("\nCLINICAL ATTRIBUTES WITH MISSING VALUES")

clinical_missing = clinical_T.isnull().sum()
print(clinical_missing[clinical_missing > 0])
```

```python
# Check duplicate feature names in each dataset

print("GENOMICS duplicate features:",
      genomics_T.columns.duplicated().sum())

print("METHYLATION duplicate features:",
      methylation_T.columns.duplicated().sum())

print("TRANSCRIPTOMICS duplicate features:",
      transcriptomics_T.columns.duplicated().sum())

print("CLINICAL duplicate attributes:",
      clinical_T.columns.duplicated().sum())
```

```python
# Check data types of each dataset

print("GENOMICS DATA TYPES")
print(genomics_T.dtypes.value_counts())

print("\nMETHYLATION DATA TYPES")
print(methylation_T.dtypes.value_counts())

print("\nTRANSCRIPTOMICS DATA TYPES")
print(transcriptomics_T.dtypes.value_counts())

print("\nCLINICAL DATA TYPES")
print(clinical_T.dtypes)
```

```python
# Check numerical value ranges

print("GENOMICS SUMMARY")
print(genomics_T.describe().loc[["min", "max", "mean"]])

print("\nMETHYLATION SUMMARY")
print(methylation_T.describe().loc[["min", "max", "mean"]])

print("\nTRANSCRIPTOMICS SUMMARY")
print(transcriptomics_T.describe().loc[["min", "max", "mean"]])
```

```python
# Inspect survival-related columns

print("OVERALL SURVIVAL:")
print(clinical_T["overall_survival"].head(15))

print("\nSTATUS:")
print(clinical_T["status"].head(15))

print("\nCOMBINED OVERALL SURVIVAL:")
print(clinical_T["overallsurvival"].head(15))
```

```python
# Convert survival time and status into numeric values

survival_time = pd.to_numeric(
    clinical_T["overall_survival"], errors="coerce"
)

survival_status = pd.to_numeric(
    clinical_T["status"], errors="coerce"
)

# Construct expected combined values

expected_combined = (
    survival_time.astype("Int64").astype("string")
    + ","
    + survival_status.astype("Int64").astype("string")
)

# Compare with existing combined column

comparison = (
    expected_combined == clinical_T["overallsurvival"].astype("string")
)

print("Matching records:", comparison.sum())
print("Non-matching records:", (~comparison).sum())
print("Missing comparisons:", comparison.isna().sum())
```

```python
# Create a copy of the aligned clinical dataset
clinical_clean = clinical_T.copy()

# Remove the redundant combined survival column
clinical_clean = clinical_clean.drop(columns=["overallsurvival"])

# Convert numeric clinical columns into numeric data types
numeric_columns = [
    "years_to_birth",
    "number_of_lymph_nodes",
    "overall_survival",
    "status"
]

for col in numeric_columns:
    clinical_clean[col] = pd.to_numeric(
        clinical_clean[col], errors="coerce"
    )

# Display the cleaned dataset
print("Cleaned clinical dataset:")
display(clinical_clean.head())

print("\nData information:")
clinical_clean.info()

print("\nDataset dimensions:", clinical_clean.shape)
```

```python
# 1. Missing values summary
print("MISSING VALUES SUMMARY")

missing_summary = pd.DataFrame({
    "Missing_Count": clinical_clean.isnull().sum(),
    "Missing_Percentage": (
        clinical_clean.isnull().sum() / len(clinical_clean) * 100
    ).round(2)
})

display(missing_summary)

# 2. Distribution of histological types
print("\nHISTOLOGICAL TYPE DISTRIBUTION")
print(clinical_clean["histological_type"].value_counts())

# 3. Distribution of gender
print("\nGENDER DISTRIBUTION")
print(clinical_clean["gender"].value_counts())

# 4. Distribution of radiation therapy
print("\nRADIATION THERAPY DISTRIBUTION")
print(clinical_clean["radiation_therapy"].value_counts())
```

```python
# Load aligned genomics dataset

output_path = "/content/drive/MyDrive/RareTumor-Omics/Preprocessed_Datasets"

genomics_clean = pd.read_csv(
    os.path.join(output_path, "Genomics_Aligned.csv"),
    index_col=0
)

print("Genomics dataset shape:", genomics_clean.shape)

print("\nFirst five patient samples:")
display(genomics_clean.head())

print("\nData information:")
genomics_clean.info()
```

```python
# Check unique values across the genomics dataset

print("Unique values in genomics dataset:")
print(np.sort(pd.unique(genomics_clean.values.ravel())))

# Check whether all values are binary
is_binary = genomics_clean.isin([0, 1]).all().all()

print("\nIs the entire dataset binary?", is_binary)

# Calculate mutation frequency for each gene
mutation_frequency = genomics_clean.sum(axis=0).sort_values(
    ascending=False
)

# Calculate mutation percentage
mutation_percentage = (
    mutation_frequency / len(genomics_clean) * 100
).round(2)

# Create summary table
mutation_summary = pd.DataFrame({
    "Mutation_Count": mutation_frequency,
    "Mutation_Percentage": mutation_percentage
})

print("\nTop 15 genes by mutation frequency:")
display(mutation_summary.head(15))
```

```python
# Calculate mutation frequency for all genes
mutation_summary = pd.DataFrame({
    "Mutation_Count": genomics_clean.sum(axis=0),
    "Mutation_Percentage": (
        genomics_clean.sum(axis=0) / len(genomics_clean) * 100
    ).round(2)
})

# Sort from highest to lowest mutation frequency
mutation_summary = mutation_summary.sort_values(
    by="Mutation_Count",
    ascending=False
)

print("Complete mutation frequency report:")
display(mutation_summary)

# Count genes with no recorded mutations
zero_mutation_genes = (
    mutation_summary["Mutation_Count"] == 0
).sum()

print("\nGenes with zero recorded mutations:", zero_mutation_genes)

# Count genes mutated in only one patient
single_mutation_genes = (
    mutation_summary["Mutation_Count"] == 1
).sum()

print("Genes mutated in only one patient:", single_mutation_genes)
```

```python
# Save mutation frequency report
mutation_summary.to_csv(
    os.path.join(
        output_path,
        "Genomics_Mutation_Frequency_Report.csv"
    )
)

# Save validated genomics dataset
genomics_clean.to_csv(
    os.path.join(
        output_path,
        "Genomics_Preprocessed.csv"
    )
)

print("Genomics preprocessing files saved successfully!")

print("\nSaved files:")
print(os.listdir(output_path))
```

```python
# Load aligned transcriptomics dataset

transcriptomics_clean = pd.read_csv(
    os.path.join(output_path, "Transcriptomics_Aligned.csv"),
    index_col=0
)

print("Transcriptomics dataset shape:",
      transcriptomics_clean.shape)

print("\nFirst five patients and ten genes:")
display(transcriptomics_clean.iloc[:5, :10])

print("\nData information:")
transcriptomics_clean.info()
```

```python
# Final summary of all four datasets

datasets = {
    "Genomics": genomics_final,
    "Epigenomics": methylation_final,
    "Transcriptomics": transcriptomics_final,
    "Clinical": clinical_final
}

summary = []

for name, df in datasets.items():
    summary.append({
        "Dataset": name,
        "Samples": df.shape[0],
        "Features": df.shape[1],
        "Missing_Values": df.isnull().sum().sum(),
        "Duplicate_Sample_IDs": df.index.duplicated().sum()
    })

summary_df = pd.DataFrame(summary)

print(summary_df.to_string(index=False))
```
